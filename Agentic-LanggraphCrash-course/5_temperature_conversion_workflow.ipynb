```python
from langgraph.graph import StateGraph, START, END
from typing import TypedDict
```

```python
class TemperatureState(TypedDict):
    celsius: float
    fahrenheit: float
```

```python
def convert_func(state: TemperatureState) -> TemperatureState:
    state["fahrenheit"] = (state["celsius"] * 9/5) + 32
    return state
```

```python
graph = StateGraph(TemperatureState)

graph.add_node("convert_to_fahrenheit", convert_func)

graph.add_edge(START, "convert_to_fahrenheit")
graph.add_edge("convert_to_fahrenheit", END)

workflow = graph.compile()
```

```python
workflow
```

```python
workflow.invoke({"celsius": 25.0})
```
